```python
from pathlib import Path
from zipfile import ZipFile

raw_path = Path(raw_file_path)

print("Archive exists:", raw_path.exists())
print("Archive size (bytes):", raw_path.stat().st_size)

with ZipFile(raw_path, "r") as archive:
    print("Files inside archive:")
    for file_info in archive.infolist():
        print(
            file_info.filename,
            "- uncompressed bytes:", file_info.file_size,
            "- compressed bytes:", file_info.compress_size
        )
```

## Acquisition Method

The verification sample was obtained from the U.S. DOT Bureau of Transportation Statistics (BTS) TranStats download interface using the T-100 International Segment (All Carriers) table.

Source reference:

https://transtats.bts.gov/DL_SelectFields.aspx?QO_fu146_anzr=Nv4Pn44vr45&gnoyr_VQ=FJE

Download selections:

- Geography: unrestricted
- Year: 2024
- Period: January
- Fields: all available fields
- Carrier scope: all carriers
- Download format received: ZIP archive containing one CSV file
- Acquisition date: 2026-09-07

The original BTS archive is preserved unchanged in:

`data/raw/T_T100I_SEGMENT_ALL_CARRIER_20260907_133912.zip`

The archive contains:

`T_T100I_SEGMENT_ALL_CARRIER.csv`

Raw files under `data/raw/` are excluded from Git tracking.

The CSV is loaded directly from the ZIP archive with Pandas, so the raw BTS archive does not need to be extracted or modified.
